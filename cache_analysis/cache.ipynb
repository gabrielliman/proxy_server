```python
import os
import glob
import numpy as np
import pandas as pd

def compare_request_times(folder_path):
    # Dictionary to store our final calculations
    results = {}
    
    # Locate all .txt files in the given folder
    file_pattern = os.path.join(folder_path, '*.txt')
    files = glob.glob(file_pattern)
    
    if not files:
        print(f"No .txt files found in the folder: {folder_path}")
        return None
        
    for file_path in files:
        # Extract just the file name without the .txt extension
        file_name = os.path.splitext(os.path.basename(file_path))[0]
        
        times = []
        with open(file_path, 'r', encoding='utf-8') as f:
            for line in f:
                # Check if the line contains our target phrase
                if "TEMPO REQUISICAO:" in line:
                    try:
                        # Split the line by ':' and grab the number part, then convert to float
                        time_val = float(line.split(":")[1].strip())
                        times.append(time_val)
                    except (IndexError, ValueError):
                        # Skip lines that can't be parsed correctly
                        continue
        
        # Calculate mean and standard deviation if we found valid data
        if times:
            mean_val = np.mean(times)
            std_val = np.std(times) # Use ddof=1 for sample standard deviation if preferred: np.std(times, ddof=1)
        else:
            mean_val = np.nan
            std_val = np.nan
            
        # Store in our results dictionary under the file name
        results[file_name] = {
            'Mean': mean_val,
            'Std Dev': std_val
        }
        
    # Convert the dictionary to a pandas DataFrame
    # By default, dictionaries are converted such that the outer keys (file names) become columns
    df = pd.DataFrame(results)
    
    return df

# --- Example Usage ---
# Replace './my_logs' with the actual path to your folder containing the text files.
folder = '.' 

comparison_df = compare_request_times(folder)

if comparison_df is not None:
    print("--- Comparison of Wiku Request Times ---")
    comparison_df.head()
    
    # Optional: Save the results to a CSV file
    # comparison_df.to_csv('comparison_results.csv')
```
